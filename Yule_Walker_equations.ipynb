```python
import numpy as np
import pandas as pd
from statsmodels.tsa.stattools import adfuller
from statsmodels.regression.linear_model import yule_walker
```

## 1) Stationarity

```python
np.random.seed(42)

sales = pd.Series(
    100 + np.random.normal(0, 5, 60)
)
```

```python
first_half = sales[:30]
second_half = sales[30:]

print("Mean (first half):", first_half.mean())
print("Mean (second half):", second_half.mean())

print("Variance (first half):", first_half.var())
print("Variance (second half):", second_half.var())
```

```python
adf_stat, p_value, *_ = adfuller(sales)

print(f"ADF Statistic: {adf_stat:.4f}")
print(f"P-value: {p_value:.4f}")
```

## 1) Autocorrelation

```python
sales = pd.Series([100, 105, 110, 108, 112])
```

```python
mu = sales.mean()
print(mu) #107
```

```python
autocov_lag1 = (
    ((sales[1:] - mu) * (sales.shift(1)[1:] - mu))
    .mean()
)
print(autocov_lag1) #4
```

```python
variance = sales.var(ddof=0)
print(variance)# 17.6
```

```python
rho_1 = autocov_lag1 / variance
print(rho_1)# 0.2557
```

## 2) Yule Walker

```python
phi, sigma = yule_walker(sales, order=2)

print(phi) #[ 0.27596224 -0.21423384]
```
